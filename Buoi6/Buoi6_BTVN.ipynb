```python
import pandas as pd
```

```python
df = pd.read_excel("Canada.xlsx", sheet_name="Canada by Citizenship",skiprows=range(20), skipfooter=2)
```

```python
df.rename(columns={
    'OdName':'Country',
    'AreaName':'Continent',
    'RegName':'Region'
}, inplace=True)
```

```python
df.drop(columns=['AREA', 'REG', 'DEV', 'Type', 'Coverage', 'DevName'], inplace=True)
```

```python
df.set_index('Country', inplace=True)
```

```python
df2 = pd.read_csv("../Buoi7/EDA_automobile.csv")
```

```python
df2['price'].plot(kind='box')
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(y=df2['price'])
```

```python
Q1 = df2['price'].quantile(0.25)
Q3 = df2['price'].quantile(0.75)

IQR = Q3 - Q1

MIN = Q1 - 1.5 * IQR
MAX = Q3 + 1.5 * IQR
```

```python
df_price_outlier = df2[
    (df2['price'] < MIN) |
    (df2['price'] > MAX)
]
df_price_outlier
```
