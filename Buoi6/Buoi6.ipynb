```python
import pandas as pd
```

```python
df = pd.read_excel("Canada.xlsx", sheet_name="Canada by Citizenship",skiprows=range(20), skipfooter=2)
```

```python
df.rename(columns={
    'OdName':'Country',
    'AreaName':'Continent',
    'RegName':'Region'
}, inplace=True)
```

```python
df.drop(columns=['AREA', 'REG', 'DEV', 'Type', 'Coverage', 'DevName'], inplace=True)
```

```python
df.set_index('Country', inplace=True)
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('ggplot')
haiti = df.loc['Haiti', 1980:2013]
```

```python
haiti.plot()
```

```python
df3 = pd.DataFrame({
    'A': [1, 2, 3.5, 4, 6],
    'B': [2, 4, 4.5, 5, 6],
    'C': [3, 5, 5.5, 6.2, 8]
})
```

```python
df3.plot(kind='area', stacked=False)
```

```python
df3.plot(kind='area', stacked=True)
```

```python
df_can = df.copy()
```

```python
df_can['Total'] = df_can[list(range(1980,2014))].sum(axis=1)
```

```python
df_can
```

```python
df_can.sort_values(by='Total', ascending=False, axis=0, inplace=True)
```

```python
df_can_top5 = df_can.head()
```

```python
df_can_top5 = df_can_top5.loc[:, 1980:2013].transpose()
```

```python
df_can_top5.plot(kind='area', stacked=False, figsize=(20, 10))
```
