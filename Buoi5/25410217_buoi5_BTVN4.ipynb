```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
df = pd.read_csv("dataset_bai2.csv")
```

```python
# Cau 1
df.replace(to_replace='?', value=np.nan, inplace=True)
df.dropna(inplace=True)
```

```python
df
```

```python
#Cau 2
cot_chuyen_sang_so = [
    "normalized-losses",
    "bore",
    "stroke",
    "horsepower",
    "peak-rpm",
    "price"
]

for cot in cot_chuyen_sang_so:
    df[cot] = LabelEncoder().fit_transform(df[cot])
```

```python
df.dtypes
```

```python
#Cau 3
#3.1
label_encoder = LabelEncoder()
df['make'] = label_encoder.fit_transform(df['make'])
```

```python
df['make']
```

```python
#3.2
thu_tu_cua = {
    "two": 2,
    "four": 4
}

df["num-of-doors"] = df["num-of-doors"].map(thu_tu_cua)
```

```python
df["num-of-doors"]
```

```python
#3.3
cot_phan_loai_con_lai = df.select_dtypes(
    include=["object"]
).columns.tolist()

df = pd.get_dummies(
    df,
    columns=cot_phan_loai_con_lai,
    dtype="int64"
)
```

```python
df
```

```python
#Cau 4
scaler = StandardScaler()
df[df.columns] = scaler.fit_transform(df[df.columns])
```
