```python
X7 = df[['horsepower', 'curb-weight']]

from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

from sklearn.pipeline import Pipeline

Input = [
    ('ss', StandardScaler()),
    ('pf7', PolynomialFeatures(degree=15, include_bias=False)),
    ('m7', LinearRegression())
]
pipe = Pipeline(Input)
pipe.fit(X7, y)
```

```python
pipe.predict(np.array([[111, 2548]]))
```

```python
yhat_m7 = pipe.predict(X7)

ax1 = sns.kdeplot(data=df, x="price")
sns.kdeplot(x=yhat_m7, ax=ax1)
```
