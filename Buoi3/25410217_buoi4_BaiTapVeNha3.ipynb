```python
import pandas as pd
import numpy as np
```

```python
# Bai 1 Xu ly du lieu khuyet thieu
```

```python
df = pd.read_csv("dataset_bai2.csv")
df
```

```python
df.replace(to_replace='?', value=np.nan, inplace=True)
```

```python
# 1. drop nan cua cot 'price'
df.dropna(axis=0, subset=['price'], inplace=True)
df
```

```python
df.dtypes
```

```python
# 2. replace nan bang mean cua cot horsepower
df['horsepower'] = pd.to_numeric(
    df['horsepower'], errors='coerce'
).astype('float64')
```

```python
horsepower_mean = df['horsepower'].mean()
df['horsepower'].replace(to_replace=np.nan, value=horsepower_mean)
df
```

```python
# 3. replace nan bang mode cua cot 'normalized-losses'
df['normalized-losses'] = pd.to_numeric(
    df['normalized-losses'], errors='coerce'
).astype('float64')
```

```python
normalized_losses_mode = df['normalized-losses'].mode()[0]
df['normalized-losses'].replace(to_replace=np.nan, value=normalized_losses_mode)
df
```

```python
#4 Loai bo tat ca cac dong co con khuyet
df.dropna(axis=0, how='any', inplace=True)
df
```

```python
#5 Chuyen kieu du lieu phu hop cho cac cot
df.dtypes
```

```python
df['price'] = pd.to_numeric(
    df['price'],
    errors='raise'
).astype('int64')

df['bore'] = pd.to_numeric(
    df['bore'],
    errors='raise'
).astype('float64')

df['stroke'] = pd.to_numeric(
    df['stroke'],
    errors='raise'
).astype('float64')

df['peak-rpm'] = pd.to_numeric(
    df['peak-rpm'],
    errors='raise'
).astype('int64')

df['normalized-losses'] = (
    df['normalized-losses']
    .round()
    .astype('int64')
)
```

```python
# Bai 2 Chuan hoa du lieu
df2 = df[['length', 'width', 'height']]
df2
```

```python
# chuan hoa scale bang cong thuc
df2['length_congthuc_scale'] = df2['length'] / df2['length'].max()
df2['width_congthuc_scale'] = df2['width'] / df2['width'].max()
df2['height_congthuc_scale'] = df2['height'] / df2['height'].max()
df2
```

```python
# chuan hoa max_min bang cong thuc
df2['length_congthuc_maxmin'] = (df2['length'] - df2['length'].min()) / (df2['length'].max() - df2['length'].min())
df2['width_congthuc_maxmin'] = (df2['width'] - df2['width'].min()) / (df2['width'].max() - df2['width'].min())
df2['height_congthuc_maxmin'] = (df2['height'] - df2['height'].min()) / (df2['height'].max() - df2['height'].min())
df2
```

```python
# chuan hoa z-score bang cong thuc
df2['length_congthuc_zscore'] = (df2['length'] - df2['length'].mean()) / df2['length'].std()
df2['width_congthuc_zscore'] = (df2['width'] - df2['width'].mean()) / df2['width'].std()
df2['height_congthuc_zscore'] = (df2['height'] - df2['height'].mean()) / df2['height'].std()
df2
```

```python
# chuan hoa scale bang sklearn
from sklearn.preprocessing import MinMaxScaler
mm = MinMaxScaler()
df2[['length_sklearn_scale']] = mm.fit_transform(df2[['length']])
df2[['width_sklearn_scale']] = mm.fit_transform(df2[['width']])
df2[['height_sklearn_scale']] = mm.fit_transform(df2[['height']])
df2
```

```python
# chuan hoa z-score bang sklearn
from sklearn.preprocessing import StandardScaler
ss = StandardScaler()
df2[['length_sklearn_zscore']] = ss.fit_transform(df2[['length']])
df2[['width_sklearn_zscore']] = ss.fit_transform(df2[['width']])
df2[['height_sklearn_zscore']] = ss.fit_transform(df2[['height']])
df2
```
