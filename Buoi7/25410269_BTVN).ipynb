```python
import pandas as pd
```

```python
df = pd.read_csv("Model_Dataset.csv")
```

```python
y = df['price']
```

```python
import seaborn as sns
```

```python
X1 = df[['horsepower', 'curb-weight', 'engine-size', 'highway-mpg']]
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline
```

```python
Input2 = [
    ('ss2', StandardScaler()),
    ('pf2', PolynomialFeatures(degree=3, include_bias=False)),
    ('m2', LinearRegression())
]
```

```python
m2 = Pipeline(Input2)
```

```python
m2.fit(X1, y)
```

```python
yhat_m2 = m2.predict(X1)
```

```python
ax2 = sns.kdeplot(data=df, x="price")
sns.kdeplot(x=yhat_m2, ax=ax2)
```

```python
Input3 = [
    ('ss3', StandardScaler()),
    ('pf3', PolynomialFeatures(degree=16, include_bias=False)),
    ('m3', LinearRegression())
]
```

```python
m3 = Pipeline(Input3)
```

```python
m3.fit(X1, y)
```

```python
yhat_m3 = m3.predict(X1)
```

```python
ax3 = sns.kdeplot(data=df, x="price")
sns.kdeplot(x=yhat_m3, ax=ax3)
```
