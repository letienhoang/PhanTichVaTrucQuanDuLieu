```python
import pandas as pd

df = pd.read_csv("EDA_automobile.csv")
```

```python
df_num = df[[
    'symboling',
    'normalized-losses',
    'wheel-base',
    'length',
    'width',
    'height',
    'curb-weight',
    'engine-size',
    'bore',
    'stroke',
    'compression-ratio',
    'horsepower',
    'peak-rpm',
    'city-mpg',
    'highway-mpg',
    'price'
]]

df_num.corr()
```

```python
corr_price = df_num.corr()['price']
corr_price
```

```python
group_4wd = df[df['drive-wheels'] == '4wd']['price']
group_fwd = df[df['drive-wheels'] == 'fwd']['price']
group_rwd = df[df['drive-wheels'] == 'rwd']['price']
```

```python
import scipy.stats as stats

f_value, p_value = stats.f_oneway(
    group_4wd,
    group_fwd,
    group_rwd
)

print("F-value:", f_value)
print("P-value:", p_value)
```

```python
if p_value < 0.05:
    print("drive-wheels có ảnh hưởng đến price")
else:
    print("drive-wheels không có ảnh hưởng đáng kể đến price")
```
